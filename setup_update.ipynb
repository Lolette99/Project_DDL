# Main Reproduction: Full FedLEASE Pipeline

<div style="background-color:#f7f7f8; padding:18px; border-radius:14px; border:1px solid #e5e5e5;">

## Objective

The goal of this part is to reproduce the main **FedLEASE** pipeline on the NLU setting described in the paper.

We aim to follow the reference experimental setup as closely as possible, using **4 GLUE tasks**, **16 clients**, **25 communication rounds**, and the adaptive LoRA expert allocation mechanism.

</div>

<br>

## Target Experimental Setup

| Element | Configuration |
|---|---|
| **Tasks** | SST-2, QNLI, MRPC, QQP |
| **Number of clients** | 16 |
| **Client distribution** | 4 clients per task |
| **Global rounds** | 25 |
| **Warmup rounds** | 5 |
| **Local epochs** | 2 |
| **LoRA rank** | 4 |
| **Mmax** | 8 |
| **Batch size** | 128 if possible |
| **Train samples** | 1000 per client |
| **Test samples** | 200 per client |

<br>

<div style="background-color:#fff8e1; padding:16px; border-radius:12px; border-left:5px solid #f5b400;">

````python
import json
from pathlib import Path
import pandas as pd

run_dir = Path("/content/FedLEASE/runs/colab_scaled_reproduction")

history_files = list(run_dir.rglob("training_history.json"))
print("Found history files:")
for f in history_files:
    print(f)

if not history_files:
    print("No training_history.json found.")
else:
    history_path = history_files[0]

    with open(history_path, "r") as f:
        history = json.load(f)

    rows = []

    for client_id, scores in history.get("client_scores", {}).items():
        if len(scores) == 0:
            continue

        final_scores = scores[-1]

        rows.append({
            "client_id": client_id,
            "final_scores": final_scores
        })

    df = pd.DataFrame(rows)
    display(df)
````

# Création d'un wrapper pour ajouter les mesures de temps.

````python
%%writefile /content/FedLEASE/run_with_timing.py
import os
import time
import json
import atexit
import statistics

import main as fedlease_main
import client as client_mod
import server as server_mod

timings = {
    "total_start": time.perf_counter(),
    "warmup_training": [],
    "clustered_training": [],
    "aggregation_warmup": [],
    "aggregation": [],
    "clustering": [],
}

def wrap_timed_method(bucket_name, original_fn):
    def wrapped(*args, **kwargs):
        start = time.perf_counter()
        try:
            return original_fn(*args, **kwargs)
        finally:
            elapsed = time.perf_counter() - start
            epochs = kwargs.get("epochs", None)
            timings[bucket_name].append({
                "seconds": elapsed,
                "epochs": epochs,
            })
    return wrapped

client_mod.WarmupClient.local_training = wrap_timed_method(
    "warmup_training",
    client_mod.WarmupClient.local_training
)

client_mod.Client.local_training = wrap_timed_method(
    "clustered_training",
    client_mod.Client.local_training
)

server_mod.Server.aggregation_warmup = wrap_timed_method(
    "aggregation_warmup",
    server_mod.Server.aggregation_warmup
)

server_mod.Server.aggregation = wrap_timed_method(
    "aggregation",
    server_mod.Server.aggregation
)

fedlease_main.compute_lora_client_map = wrap_timed_method(
    "clustering",
    fedlease_main.compute_lora_client_map
)

def summarize_bucket(values):
    seconds = [v["seconds"] for v in values]
    per_epoch = [
        v["seconds"] / v["epochs"]
        for v in values
        if v.get("epochs")
    ]

    return {
        "calls": len(values),
        "total_seconds": sum(seconds) if seconds else 0.0,
        "mean_seconds": statistics.mean(seconds) if seconds else 0.0,
        "mean_seconds_per_epoch": statistics.mean(per_epoch) if per_epoch else None,
    }

def save_timings():
    total_seconds = time.perf_counter() - timings["total_start"]

    summary = {
        "total_training_time_seconds": total_seconds,
        "warmup_training": summarize_bucket(timings["warmup_training"]),
        "clustered_training": summarize_bucket(timings["clustered_training"]),
        "aggregation_warmup": summarize_bucket(timings["aggregation_warmup"]),
        "aggregation": summarize_bucket(timings["aggregation"]),
        "clustering": summarize_bucket(timings["clustering"]),
        "raw": timings,
    }

    timing_file = os.environ.get(
        "TIMING_FILE",
        "/content/FedLEASE/timing_summary.json"
    )

    with open(timing_file, "w") as f:
        json.dump(summary, f, indent=2)

    print(f"\nTiming summary saved to: {timing_file}")

atexit.register(save_timings)

if __name__ == "__main__":
    fedlease_main.main()
````
